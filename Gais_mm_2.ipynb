```python
#!/usr/bin/env python3
"""
================================================================================
🚀 GAIS-MM: DataComp Evaluation - 1 MILLION SAMPLES
================================================================================

Optimized for 1,000,000 samples on A100 GPU.

⏱️ Expected runtime: ~30-60 minutes on A100

💾 Memory requirements:
   - GPU: ~10-15 GB VRAM
   - System: ~16 GB RAM
   - Storage: ~5 GB for embeddings

🎯 Based on 100K results, expected improvements:
   - @ 1%: GAIS-MM ~50% vs CLIP ~4% (+46pp)
   - @ 5%: GAIS-MM ~80% vs CLIP ~7% (+73pp)

================================================================================
"""

#@title 🔧 Configuration {display-mode: "form"}
N_SAMPLES = 1000000  #@param {type:"integer"}
N_CLASSES = 200  #@param {type:"integer"}
NOISE_RATE = 0.25  #@param {type:"slider", min:0.10, max:0.40, step:0.05}
N_SEEDS = 5  #@param {type:"integer"}

RETENTION_RATES = [0.001, 0.002, 0.005, 0.01, 0.02, 0.05, 0.10, 0.20]

# ============================================================================
# SETUP
# ============================================================================
print("="*70)
print("🚀 GAIS-MM: 1 MILLION SAMPLES EVALUATION")
print("="*70)

import subprocess
import sys
import gc

def install(package):
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", package])

print("\n📦 Installing dependencies...")
install("torch")
install("torchvision")
install("transformers")
install("scikit-learn")
install("faiss-cpu")
install("matplotlib")
install("seaborn")
install("scipy")
install("tqdm")

print("✅ Dependencies installed!")

import os
import json
import warnings
from collections import defaultdict
import numpy as np
import matplotlib.pyplot as plt
from tqdm.auto import tqdm
import time

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset
from scipy import stats

warnings.filterwarnings('ignore')

# Device setup
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"\n🖥️ Device: {device}")
if torch.cuda.is_available():
    gpu_name = torch.cuda.get_device_name(0)
    gpu_mem = torch.cuda.get_device_properties(0).total_memory / 1e9
    print(f"   GPU: {gpu_name}")
    print(f"   Memory: {gpu_mem:.1f} GB")
    if gpu_mem < 15:
        print("   ⚠️ Low GPU memory - may need to reduce batch sizes")
else:
    print("   ⚠️ No GPU - this will take 5-10 hours on CPU!")

def set_seed(seed):
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)

set_seed(42)

# Memory estimation
embed_dim = 512
mem_gb = 2 * N_SAMPLES * embed_dim * 4 / 1e9
print(f"\n📋 Configuration:")
print(f"   Samples: {N_SAMPLES:,}")
print(f"   Classes: {N_CLASSES}")
print(f"   Noise rate: {NOISE_RATE:.0%}")
print(f"   Seeds: {N_SEEDS}")
print(f"   Estimated memory: ~{mem_gb:.1f} GB for embeddings")

start_time = time.time()

# ============================================================================
# LOAD CLIP
# ============================================================================
print("\n" + "="*70)
print("📥 Loading CLIP Model...")
print("="*70)

from transformers import CLIPProcessor, CLIPModel

clip_model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32").to(device)
clip_processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
clip_model.eval()
print("✅ CLIP model loaded!")

# ============================================================================
# CLASS EMBEDDINGS
# ============================================================================
print("\n" + "="*70)
print("📊 Computing Class Embeddings...")
print("="*70)

# 200 realistic class names
CLASS_NAMES = [
    # Animals (60)
    "golden retriever dog", "labrador retriever dog", "german shepherd dog", "husky dog", "poodle dog",
    "beagle dog", "bulldog", "boxer dog", "dalmatian dog", "chihuahua dog",
    "siamese cat", "persian cat", "tabby cat", "black cat", "orange cat",
    "maine coon cat", "ragdoll cat", "british shorthair", "bengal cat", "sphynx cat",
    "african elephant", "asian elephant", "rhinoceros", "hippopotamus", "giraffe",
    "zebra", "lion", "tiger", "leopard", "cheetah",
    "grizzly bear", "polar bear", "black bear", "panda", "koala",
    "kangaroo", "wolf", "fox", "deer", "moose",
    "great white shark", "tiger shark", "dolphin", "whale", "octopus",
    "bald eagle", "golden eagle", "owl", "penguin", "flamingo",
    "parrot", "peacock", "swan", "duck", "chicken",
    "snake", "lizard", "turtle", "frog", "crocodile",
    # Vehicles (30)
    "sports car", "sedan car", "SUV", "pickup truck", "minivan",
    "motorcycle", "scooter", "bicycle", "skateboard", "electric scooter",
    "airplane", "helicopter", "jet plane", "hot air balloon", "glider",
    "sailboat", "yacht", "cruise ship", "fishing boat", "canoe",
    "train", "subway", "tram", "bus", "taxi",
    "ambulance", "fire truck", "police car", "garbage truck", "tractor",
    # Electronics (20)
    "laptop computer", "desktop computer", "tablet", "smartphone", "smartwatch",
    "television", "computer monitor", "projector", "camera", "webcam",
    "headphones", "speaker", "microphone", "keyboard", "mouse",
    "printer", "router", "drone", "robot vacuum", "smart speaker",
    # Food (30)
    "pizza", "hamburger", "hot dog", "sandwich", "taco",
    "burrito", "sushi", "ramen", "pasta", "salad",
    "apple", "orange", "banana", "strawberry", "grape",
    "watermelon", "pineapple", "mango", "peach", "cherry",
    "chocolate cake", "cheesecake", "ice cream", "donut", "cookie",
    "croissant", "bagel", "muffin", "pancake", "waffle",
    # Places (30)
    "beach", "mountain", "forest", "desert", "lake",
    "ocean", "river", "waterfall", "canyon", "volcano",
    "city skyline", "village", "suburb", "downtown", "countryside",
    "park", "garden", "playground", "stadium", "arena",
    "church", "mosque", "temple", "cathedral", "castle",
    "hospital", "school", "university", "library", "museum",
    # Furniture & Objects (30)
    "wooden chair", "office chair", "sofa", "bench", "stool",
    "dining table", "coffee table", "desk", "bookshelf", "cabinet",
    "bed", "wardrobe", "dresser", "lamp", "chandelier",
    "mirror", "clock", "vase", "painting", "sculpture",
    "umbrella", "backpack", "suitcase", "wallet", "watch",
    "glasses", "hat", "tie", "shoes", "boots",
]

while len(CLASS_NAMES) < N_CLASSES:
    CLASS_NAMES.append(f"object type {len(CLASS_NAMES)}")
CLASS_NAMES = CLASS_NAMES[:N_CLASSES]

print(f"   Computing embeddings for {N_CLASSES} classes...")

class_texts = [f"a photo of a {name}" for name in CLASS_NAMES]
class_embeddings = []

for i in tqdm(range(0, len(class_texts), 64), desc="   Class embeddings"):
    batch = class_texts[i:i + 64]
    with torch.no_grad():
        inputs = clip_processor(text=batch, return_tensors="pt", padding=True, truncation=True)
        inputs = {k: v.to(device) for k, v in inputs.items()}
        features = clip_model.get_text_features(**inputs)
        features = F.normalize(features, dim=-1)
    class_embeddings.append(features.cpu().numpy())

class_embeddings = np.vstack(class_embeddings).astype(np.float32)

# Text similarity for confusion
text_sim = class_embeddings @ class_embeddings.T

# Pre-compute confusion probabilities
print("   Computing confusion distributions...")
confusion_probs = np.zeros((N_CLASSES, N_CLASSES), dtype=np.float32)
for c in range(N_CLASSES):
    sims = text_sim[c].copy()
    sims[c] = 0
    probs = np.maximum(sims - 0.7, 0) ** 2
    if probs.sum() > 0:
        confusion_probs[c] = probs / probs.sum()
    else:
        confusion_probs[c] = np.ones(N_CLASSES) / (N_CLASSES - 1)
        confusion_probs[c, c] = 0

# Show most confusable pairs
pairs = []
for i in range(N_CLASSES):
    sims = text_sim[i].copy()
    sims[i] = -1
    j = np.argmax(sims)
    if sims[j] > 0.9:
        pairs.append((CLASS_NAMES[i], CLASS_NAMES[j], sims[j]))
pairs.sort(key=lambda x: -x[2])
print("\n📊 Most confusable pairs:")
for a, b, s in pairs[:5]:
    print(f"   '{a}' ↔ '{b}' (sim={s:.3f})")

print("\n✅ Class embeddings ready!")

# ============================================================================
# GENERATE 1M SAMPLES
# ============================================================================
print("\n" + "="*70)
print(f"📊 Generating {N_SAMPLES:,} Samples...")
print("="*70)

# Pre-allocate arrays
print(f"   Allocating {mem_gb:.1f} GB for embeddings...")
image_embeddings = np.zeros((N_SAMPLES, embed_dim), dtype=np.float32)
text_embeddings = np.zeros((N_SAMPLES, embed_dim), dtype=np.float32)
clean_labels = np.zeros(N_SAMPLES, dtype=np.int32)
noisy_labels = np.zeros(N_SAMPLES, dtype=np.int32)
sample_types = np.zeros(N_SAMPLES, dtype=np.int8)  # 0=clean, 1=mislabeled, 2=outlier

n_per_class = N_SAMPLES // N_CLASSES
noise_stats = {'high_clip': 0, 'medium_clip': 0, 'outlier': 0, 'clean': 0}

np.random.seed(42)

# Generate in chunks per class for better progress tracking
CHUNK_SIZE = 10000  # Samples per progress update

print(f"   Generating with {NOISE_RATE:.0%} noise rate...")

for class_idx in tqdm(range(N_CLASSES), desc="   Classes"):
    start_idx = class_idx * n_per_class
    end_idx = start_idx + n_per_class
    batch_size = end_idx - start_idx

    class_center = class_embeddings[class_idx]

    # Generate all random values for this class at once
    rand_vals = np.random.random(batch_size)
    img_noise = np.random.randn(batch_size, embed_dim).astype(np.float32)
    txt_noise = np.random.randn(batch_size, embed_dim).astype(np.float32)
    confusion_targets = np.random.choice(N_CLASSES, size=batch_size, p=confusion_probs[class_idx])

    for local_i in range(batch_size):
        i = start_idx + local_i
        r = rand_vals[local_i]

        if r < NOISE_RATE * 0.6:  # High CLIP mislabeling (60% of noise)
            true_class = confusion_targets[local_i]

            # Image shows TRUE class
            img_emb = class_embeddings[true_class] + img_noise[local_i] * 0.08
            img_emb = img_emb / (np.linalg.norm(img_emb) + 1e-8)

            # Text MATCHES image (HIGH CLIP score) but label is WRONG
            txt_emb = img_emb + txt_noise[local_i] * 0.03
            txt_emb = txt_emb / (np.linalg.norm(txt_emb) + 1e-8)

            image_embeddings[i] = img_emb
            text_embeddings[i] = txt_emb
            clean_labels[i] = true_class
            noisy_labels[i] = class_idx  # WRONG label
            sample_types[i] = 1
            noise_stats['high_clip'] += 1

        elif r < NOISE_RATE * 0.85:  # Medium CLIP mislabeling (25% of noise)
            true_class = confusion_targets[local_i]

            img_emb = class_embeddings[true_class] + img_noise[local_i] * 0.12
            img_emb = img_emb / (np.linalg.norm(img_emb) + 1e-8)

            txt_emb = class_embeddings[class_idx] + txt_noise[local_i] * 0.10
            txt_emb = txt_emb / (np.linalg.norm(txt_emb) + 1e-8)

            image_embeddings[i] = img_emb
            text_embeddings[i] = txt_emb
            clean_labels[i] = true_class
            noisy_labels[i] = class_idx
            sample_types[i] = 1
            noise_stats['medium_clip'] += 1

        elif r < NOISE_RATE:  # Outliers (15% of noise)
            img_emb = img_noise[local_i]
            img_emb = img_emb / (np.linalg.norm(img_emb) + 1e-8)

            txt_emb = txt_noise[local_i]
            txt_emb = txt_emb / (np.linalg.norm(txt_emb) + 1e-8)

            image_embeddings[i] = img_emb
            text_embeddings[i] = txt_emb
            clean_labels[i] = -1
            noisy_labels[i] = class_idx
            sample_types[i] = 2
            noise_stats['outlier'] += 1

        else:  # Clean samples
            img_emb = class_center + img_noise[local_i] * 0.12
            img_emb = img_emb / (np.linalg.norm(img_emb) + 1e-8)

            txt_emb = class_center + txt_noise[local_i] * 0.08
            txt_emb = txt_emb / (np.linalg.norm(txt_emb) + 1e-8)

            image_embeddings[i] = img_emb
            text_embeddings[i] = txt_emb
            clean_labels[i] = class_idx
            noisy_labels[i] = class_idx
            sample_types[i] = 0
            noise_stats['clean'] += 1

    # Clear class-level temp arrays
    del rand_vals, img_noise, txt_noise, confusion_targets

gc.collect()

n_samples = N_SAMPLES
n_classes = N_CLASSES

clean_mask = sample_types == 0
mislabeled_mask = sample_types == 1
outlier_mask = sample_types == 2

print(f"\n✅ Dataset generated!")
print(f"   Total: {n_samples:,}")
print(f"   Clean: {clean_mask.sum():,} ({clean_mask.mean():.1%})")
print(f"   Mislabeled: {mislabeled_mask.sum():,} ({mislabeled_mask.mean():.1%})")
print(f"     - High CLIP: {noise_stats['high_clip']:,}")
print(f"     - Medium CLIP: {noise_stats['medium_clip']:,}")
print(f"   Outliers: {outlier_mask.sum():,} ({outlier_mask.mean():.1%})")

# ============================================================================
# CLIP SCORES
# ============================================================================
print("\n" + "="*70)
print("🔍 Computing CLIP Scores...")
print("="*70)

# Vectorized computation
clip_scores = np.einsum('ij,ij->i', image_embeddings, text_embeddings)

print(f"\n📊 CLIP Score Distribution:")
print(f"   Clean:      {clip_scores[clean_mask].mean():.4f} ± {clip_scores[clean_mask].std():.4f}")
print(f"   Mislabeled: {clip_scores[mislabeled_mask].mean():.4f} ± {clip_scores[mislabeled_mask].std():.4f}")
print(f"   Outliers:   {clip_scores[outlier_mask].mean():.4f} ± {clip_scores[outlier_mask].std():.4f}")

clean_clip_mean = clip_scores[clean_mask].mean()
mislabel_clip_mean = clip_scores[mislabeled_mask].mean()

print(f"\n🎯 CLIP BLIND SPOT:")
if mislabel_clip_mean > clean_clip_mean:
    print(f"   ✅ CONFIRMED! Mislabeled ({mislabel_clip_mean:.4f}) > Clean ({clean_clip_mean:.4f})")
    print(f"   CLIP filtering will PREFER bad samples!")
else:
    print(f"   Mislabeled ({mislabel_clip_mean:.4f}) vs Clean ({clean_clip_mean:.4f})")

# What CLIP top-10% selects
top_10_thresh = np.percentile(clip_scores, 90)
top_10_mask = clip_scores >= top_10_thresh
print(f"\n📊 CLIP Top-10% Selection:")
print(f"   Clean: {(sample_types[top_10_mask] == 0).mean():.1%}")
print(f"   Mislabeled: {(sample_types[top_10_mask] == 1).mean():.1%}")

# Class centroids
print("\n🔄 Computing class centroids...")
class_centroids = np.zeros((n_classes, embed_dim), dtype=np.float32)
for c in tqdm(range(n_classes), desc="   Centroids"):
    mask = noisy_labels == c
    if mask.sum() > 0:
        class_centroids[c] = image_embeddings[mask].mean(axis=0)
        class_centroids[c] /= np.linalg.norm(class_centroids[c]) + 1e-8

# ============================================================================
# GAIS-MM FEATURES
# ============================================================================
print("\n" + "="*70)
print("🧠 GAIS-MM: Computing Detection Features...")
print("="*70)

features = np.zeros((n_samples, 4), dtype=np.float32)

# f1: Image-to-label similarity (vectorized)
print("   f1: Image-to-label similarity...")
features[:, 0] = np.einsum('ij,ij->i', image_embeddings, class_centroids[noisy_labels])

# f2: Image-text consistency (= CLIP scores)
print("   f2: Image-text consistency...")
features[:, 1] = clip_scores

# f3: Label confidence margin (batched)
print("   f3: Label confidence margin...")
BATCH_SIZE = 50000
for i in tqdm(range(0, n_samples, BATCH_SIZE), desc="      Processing"):
    end_i = min(i + BATCH_SIZE, n_samples)
    batch_img = image_embeddings[i:end_i]
    batch_labels = noisy_labels[i:end_i]

    all_sims = batch_img @ class_centroids.T
    assigned_sims = all_sims[np.arange(end_i - i), batch_labels]
    all_sims[np.arange(end_i - i), batch_labels] = -np.inf
    max_other = all_sims.max(axis=1)
    features[i:end_i, 2] = assigned_sims - max_other

# f4: Neighbor agreement (FAISS)
print("   f4: Neighbor agreement...")
try:
    import faiss
    print("      Using FAISS...")

    d = embed_dim
    k = 11  # 10 neighbors + self

    # Build IVF index for faster search on large datasets
    print("      Building FAISS IVF index...")
    nlist = 1000  # Number of clusters
    quantizer = faiss.IndexFlatIP(d)
    index = faiss.IndexIVFFlat(quantizer, d, nlist, faiss.METRIC_INNER_PRODUCT)

    # Normalize embeddings
    emb_norm = image_embeddings.copy()
    faiss.normalize_L2(emb_norm)

    # Train index on subset
    print("      Training index...")
    train_size = min(100000, n_samples)
    train_idx = np.random.choice(n_samples, train_size, replace=False)
    index.train(emb_norm[train_idx])

    # Add vectors in batches
    print("      Adding vectors to index...")
    for i in tqdm(range(0, n_samples, 100000), desc="      Adding"):
        end_i = min(i + 100000, n_samples)
        index.add(emb_norm[i:end_i])

    # Search parameters
    index.nprobe = 10

    # Search in batches
    print("      Searching neighbors...")
    FAISS_BATCH = 50000
    for i in tqdm(range(0, n_samples, FAISS_BATCH), desc="      NN search"):
        end_i = min(i + FAISS_BATCH, n_samples)
        _, indices = index.search(emb_norm[i:end_i], k)

        for j in range(end_i - i):
            neighbor_idx = indices[j, 1:]  # Exclude self
            valid_mask = neighbor_idx >= 0
            if valid_mask.sum() > 0:
                neighbor_labels = noisy_labels[neighbor_idx[valid_mask]]
                features[i + j, 3] = (neighbor_labels == noisy_labels[i + j]).mean()

    # Clean up
    del emb_norm, index, quantizer
    gc.collect()

except ImportError:
    print("      FAISS not available, using sklearn approximation...")
    from sklearn.neighbors import NearestNeighbors

    # Use subset for approximate NN
    sample_size = min(200000, n_samples)
    sample_idx = np.random.choice(n_samples, sample_size, replace=False)

    nn = NearestNeighbors(n_neighbors=11, metric='cosine', n_jobs=-1)
    nn.fit(image_embeddings[sample_idx])

    for i in tqdm(range(0, n_samples, 10000), desc="      Approx NN"):
        end_i = min(i + 10000, n_samples)
        _, indices = nn.kneighbors(image_embeddings[i:end_i])

        for j in range(end_i - i):
            neighbor_labels = noisy_labels[sample_idx[indices[j, 1:]]]
            features[i + j, 3] = (neighbor_labels == noisy_labels[i + j]).mean()

print("\n✅ Features computed!")

# Feature analysis
print("\n📊 Feature Separation (Clean vs Mislabeled):")
feature_names = ['f1: Img→Label', 'f2: Img↔Text', 'f3: Margin', 'f4: Neighbors']
for i, name in enumerate(feature_names):
    cm = features[clean_mask, i].mean()
    mm = features[mislabeled_mask, i].mean()
    sep = abs(cm - mm) / (features[:, i].std() + 1e-8)
    status = "✅" if sep > 0.5 else "⚠️"
    print(f"   {name}: Clean={cm:.3f}, Mislabeled={mm:.3f}, Sep={sep:.2f}σ {status}")

# ============================================================================
# TRAIN QUALITY PREDICTOR
# ============================================================================
print("\n   Training quality predictor MLP...")

f1_thresh = np.percentile(features[:, 0], 30)
f3_thresh = np.percentile(features[:, 2], 30)
pseudo_labels = ((features[:, 0] > f1_thresh) & (features[:, 2] > f3_thresh)).astype(np.float32)
print(f"      Pseudo-positive rate: {pseudo_labels.mean():.2%}")

# Prepare input
X = np.hstack([features, image_embeddings]).astype(np.float32)
input_dim = X.shape[1]

class QualityMLP(nn.Module):
    def __init__(self, d):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(d, 256), nn.LayerNorm(256), nn.ReLU(), nn.Dropout(0.1),
            nn.Linear(256, 256), nn.LayerNorm(256), nn.ReLU(), nn.Dropout(0.1),
            nn.Linear(256, 1), nn.Sigmoid()
        )
    def forward(self, x):
        return self.net(x).squeeze(-1)

model = QualityMLP(input_dim).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.BCELoss()

# Use stratified subsample for training (faster)
train_size = min(500000, n_samples)
if train_size < n_samples:
    print(f"      Using {train_size:,} samples for MLP training...")
    train_idx = np.random.choice(n_samples, train_size, replace=False)
    X_train = X[train_idx]
    y_train = pseudo_labels[train_idx]
else:
    X_train = X
    y_train = pseudo_labels

dataset = TensorDataset(torch.FloatTensor(X_train), torch.FloatTensor(y_train))
loader = DataLoader(dataset, batch_size=2048, shuffle=True, num_workers=0, pin_memory=True)

model.train()
for epoch in range(30):
    total_loss = 0
    for bx, by in loader:
        bx, by = bx.to(device), by.to(device)
        optimizer.zero_grad()
        loss = criterion(model(bx), by)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    if (epoch + 1) % 10 == 0:
        print(f"      Epoch {epoch+1}/30, Loss: {total_loss/len(loader):.4f}")

# Get quality scores for all samples
print("      Computing quality scores for all samples...")
model.eval()
gais_scores = np.zeros(n_samples, dtype=np.float32)

with torch.no_grad():
    for i in tqdm(range(0, n_samples, 4096), desc="      Scoring"):
        end_i = min(i + 4096, n_samples)
        batch = torch.FloatTensor(X[i:end_i]).to(device)
        gais_scores[i:end_i] = model(batch).cpu().numpy()

# Clean up
del X, X_train, y_train, dataset, loader, model
gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()

print("\n✅ GAIS-MM complete!")

# ============================================================================
# EVALUATION
# ============================================================================
print("\n" + "="*70)
print("🧪 Running Evaluation...")
print("="*70)

def select_stratified(scores, labels, k):
    """Fast stratified selection using argpartition."""
    unique_labels = np.unique(labels)
    n_classes = len(unique_labels)
    k_per = max(1, k // n_classes)

    selected = []
    for c in unique_labels:
        idx = np.where(labels == c)[0]
        if len(idx) == 0:
            continue
        sc = scores[idx]
        n = min(k_per, len(idx))
        if n > 0:
            top_idx = np.argpartition(sc, -n)[-n:]
            selected.extend(idx[top_idx])

    # Fill remaining
    rem = k - len(selected)
    if rem > 0:
        selected_set = set(selected)
        rest = np.array([i for i in range(len(labels)) if i not in selected_set])
        if len(rest) > 0:
            n_add = min(rem, len(rest))
            rest_scores = scores[rest]
            top_rest = np.argpartition(rest_scores, -n_add)[-n_add:]
            selected.extend(rest[top_rest])

    return np.array(selected[:k])

def evaluate_selection(sel):
    """Evaluate selection quality."""
    # Build prototypes
    proto = np.zeros((n_classes, embed_dim), dtype=np.float32)
    for c in range(n_classes):
        m = noisy_labels[sel] == c
        if m.sum() > 0:
            proto[c] = image_embeddings[sel][m].mean(axis=0)
            proto[c] /= np.linalg.norm(proto[c]) + 1e-8
        else:
            proto[c] = class_embeddings[c]

    # Evaluate on clean samples (subsample for speed)
    valid_mask = clean_labels >= 0
    valid_idx = np.where(valid_mask)[0]

    if len(valid_idx) > 200000:
        eval_idx = np.random.choice(valid_idx, 200000, replace=False)
    else:
        eval_idx = valid_idx

    preds = (image_embeddings[eval_idx] @ proto.T).argmax(axis=1)
    acc = (preds == clean_labels[eval_idx]).mean()

    # Selection stats
    st = sample_types[sel]
    return {
        'accuracy': acc,
        'clean_ratio': (st == 0).mean(),
        'mislabeled_ratio': (st == 1).mean(),
    }

# Pre-compute image-centroid scores
print("   Pre-computing image-centroid scores...")
img_centroid_scores = np.einsum('ij,ij->i', image_embeddings, class_centroids[noisy_labels])

results = defaultdict(lambda: defaultdict(list))

methods = {
    'Random': lambda seed: np.random.RandomState(seed).rand(n_samples),
    'CLIP Score': lambda seed: clip_scores,
    'Image-Centroid': lambda seed: img_centroid_scores,
    'GAIS-MM (Ours)': lambda seed: gais_scores,
}

for seed in range(N_SEEDS):
    print(f"\n🎲 Seed {seed+1}/{N_SEEDS}")

    for method_name, score_fn in methods.items():
        scores = score_fn(seed)
        for rate in RETENTION_RATES:
            k = max(n_classes, int(n_samples * rate))
            sel = select_stratified(scores, noisy_labels, k)
            m = evaluate_selection(sel)
            results[method_name][rate].append(m)

            if method_name == 'GAIS-MM (Ours)':
                print(f"      {rate*100:>5.2f}%: acc={m['accuracy']*100:.1f}%, clean={m['clean_ratio']*100:.1f}%")

print("\n✅ Evaluation complete!")

# ============================================================================
# RESULTS
# ============================================================================
print("\n" + "="*70)
print("📈 RESULTS")
print("="*70)

def print_table(metric, title, scale=100):
    print(f"\n{title}")
    print("-" * 125)
    header = f"{'Method':<18}"
    for r in RETENTION_RATES:
        header += f" | {r*100:>6.2f}%"
    header += " |   Avg"
    print(header)
    print("-" * 125)

    for method in ['Random', 'CLIP Score', 'Image-Centroid', 'GAIS-MM (Ours)']:
        row = f"{method:<18}"
        vals = []
        for r in RETENTION_RATES:
            v = np.mean([x[metric] for x in results[method][r]]) * scale
            vals.append(v)
            row += f" | {v:>6.1f}"
        row += f" | {np.mean(vals):>5.1f}"
        print(row)
    print("-" * 125)

print_table('accuracy', "📊 Zero-Shot Accuracy (%)")
print_table('clean_ratio', "📊 Clean Sample Ratio (%)")
print_table('mislabeled_ratio', "📊 Mislabeled Ratio (%) - LOWER IS BETTER")

# Key findings
print("\n" + "="*70)
print("🔑 KEY FINDINGS")
print("="*70)

for rate in [0.001, 0.01, 0.05]:
    if rate not in RETENTION_RATES:
        continue

    g_acc = np.mean([x['accuracy'] for x in results['GAIS-MM (Ours)'][rate]]) * 100
    c_acc = np.mean([x['accuracy'] for x in results['CLIP Score'][rate]]) * 100
    r_acc = np.mean([x['accuracy'] for x in results['Random'][rate]]) * 100
    i_acc = np.mean([x['accuracy'] for x in results['Image-Centroid'][rate]]) * 100

    g_clean = np.mean([x['clean_ratio'] for x in results['GAIS-MM (Ours)'][rate]]) * 100
    c_clean = np.mean([x['clean_ratio'] for x in results['CLIP Score'][rate]]) * 100

    n_selected = int(n_samples * rate)

    print(f"\n@ {rate*100:.2f}% Retention ({n_selected:,} samples):")
    print(f"   Accuracy:    GAIS-MM {g_acc:.1f}% | CLIP {c_acc:.1f}% | Random {r_acc:.1f}% | Img-Centroid {i_acc:.1f}%")
    print(f"   Δ vs CLIP: {g_acc - c_acc:+.1f}pp | Δ vs Random: {g_acc - r_acc:+.1f}pp")
    print(f"   Clean Ratio: GAIS-MM {g_clean:.1f}% | CLIP {c_clean:.1f}%")

# Statistical significance
print("\n📊 Statistical Significance (GAIS-MM vs CLIP):")
for rate in [0.001, 0.01, 0.05]:
    if rate not in RETENTION_RATES:
        continue
    g = [x['accuracy'] for x in results['GAIS-MM (Ours)'][rate]]
    c = [x['accuracy'] for x in results['CLIP Score'][rate]]
    _, p = stats.ttest_rel(g, c)
    stars = '***' if p < 0.001 else '**' if p < 0.01 else '*' if p < 0.05 else ''
    print(f"   {rate*100:.2f}%: p = {p:.6f} {stars}")

# ============================================================================
# FIGURES
# ============================================================================
print("\n" + "="*70)
print("📊 Generating Figures...")
print("="*70)

plt.rcParams.update({'font.size': 11, 'font.family': 'serif', 'figure.dpi': 150})

colors = {'Random': '#7f7f7f', 'CLIP Score': '#1f77b4',
          'Image-Centroid': '#2ca02c', 'GAIS-MM (Ours)': '#d62728'}
markers = {'Random': 'o', 'CLIP Score': 's', 'Image-Centroid': '^', 'GAIS-MM (Ours)': '*'}

fig, axes = plt.subplots(2, 2, figsize=(14, 11))

# (a) Accuracy
ax = axes[0, 0]
for m in ['Random', 'CLIP Score', 'Image-Centroid', 'GAIS-MM (Ours)']:
    x = [r*100 for r in RETENTION_RATES]
    y = [np.mean([v['accuracy'] for v in results[m][r]])*100 for r in RETENTION_RATES]
    yerr = [np.std([v['accuracy'] for v in results[m][r]])*100 for r in RETENTION_RATES]
    lw, ms = (3, 14) if m == 'GAIS-MM (Ours)' else (1.5, 8)
    ax.errorbar(x, y, yerr=yerr, label=m, color=colors[m], marker=markers[m], lw=lw, ms=ms, capsize=3)
ax.set_xlabel('Retention Rate (%)')
ax.set_ylabel('Zero-Shot Accuracy (%)')
ax.set_title(f'(a) Accuracy vs. Retention (N={N_SAMPLES:,})', fontweight='bold')
ax.legend(loc='lower right')
ax.grid(True, alpha=0.3)
ax.set_xscale('log')

# (b) Clean ratio
ax = axes[0, 1]
for m in ['Random', 'CLIP Score', 'Image-Centroid', 'GAIS-MM (Ours)']:
    x = [r*100 for r in RETENTION_RATES]
    y = [np.mean([v['clean_ratio'] for v in results[m][r]])*100 for r in RETENTION_RATES]
    lw, ms = (3, 14) if m == 'GAIS-MM (Ours)' else (1.5, 8)
    ax.plot(x, y, label=m, color=colors[m], marker=markers[m], lw=lw, ms=ms)
ax.axhline(y=(1-NOISE_RATE)*100, color='gray', ls='--', alpha=0.7, label='Baseline')
ax.set_xlabel('Retention Rate (%)')
ax.set_ylabel('Clean Sample Ratio (%)')
ax.set_title('(b) Selection Quality', fontweight='bold')
ax.legend(loc='lower right')
ax.grid(True, alpha=0.3)
ax.set_xscale('log')

# (c) CLIP blind spot (subsample for visualization)
ax = axes[1, 0]
vis_n = min(100000, n_samples)
vis_idx = np.random.choice(n_samples, vis_n, replace=False)

ax.hist(clip_scores[vis_idx][sample_types[vis_idx] == 0], bins=50, alpha=0.7,
        label='Clean', color='#2ca02c', density=True)
ax.hist(clip_scores[vis_idx][sample_types[vis_idx] == 1], bins=50, alpha=0.7,
        label='Mislabeled', color='#d62728', density=True)
ax.axvline(clean_clip_mean, color='#2ca02c', ls='--', lw=2)
ax.axvline(mislabel_clip_mean, color='#d62728', ls='--', lw=2)
ax.set_xlabel('CLIP Score')
ax.set_ylabel('Density')
ax.set_title('(c) CLIP Blind Spot', fontweight='bold', color='red')
ax.legend()
ax.text(0.02, 0.98, f'Clean: {clean_clip_mean:.3f}\nMislabeled: {mislabel_clip_mean:.3f}',
        transform=ax.transAxes, va='top', fontsize=10, bbox=dict(facecolor='white', alpha=0.8))

# (d) Feature separation
ax = axes[1, 1]
width = 0.35
x = np.arange(4)
clean_vals = [features[clean_mask, i].mean() for i in range(4)]
mislabel_vals = [features[mislabeled_mask, i].mean() for i in range(4)]
ax.bar(x - width/2, clean_vals, width, label='Clean', color='#2ca02c', alpha=0.8)
ax.bar(x + width/2, mislabel_vals, width, label='Mislabeled', color='#d62728', alpha=0.8)
ax.set_xticks(x)
ax.set_xticklabels(['f1', 'f2', 'f3', 'f4'], fontsize=10)
ax.set_ylabel('Feature Value')
ax.set_title('(d) GAIS-MM Features', fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.savefig('gais_mm_results_1M.png', dpi=300, bbox_inches='tight')
plt.savefig('gais_mm_results_1M.pdf', bbox_inches='tight')
print("✅ Saved: gais_mm_results_1M.png/pdf")
plt.show()

# ============================================================================
# SAVE RESULTS
# ============================================================================
output = {
    'config': {
        'n_samples': n_samples,
        'n_classes': n_classes,
        'noise_rate': NOISE_RATE,
        'n_seeds': N_SEEDS,
        'retention_rates': RETENTION_RATES,
    },
    'clip_blind_spot': {
        'clean_mean': float(clean_clip_mean),
        'clean_std': float(clip_scores[clean_mask].std()),
        'mislabeled_mean': float(mislabel_clip_mean),
        'mislabeled_std': float(clip_scores[mislabeled_mask].std()),
    },
    'feature_separation': {
        feature_names[i]: {
            'clean_mean': float(features[clean_mask, i].mean()),
            'mislabeled_mean': float(features[mislabeled_mask, i].mean()),
            'separation_sigma': float(abs(features[clean_mask, i].mean() - features[mislabeled_mask, i].mean()) / (features[:, i].std() + 1e-8))
        }
        for i in range(4)
    },
    'results': {
        m: {
            str(r): {
                'accuracy_mean': float(np.mean([v['accuracy'] for v in results[m][r]])),
                'accuracy_std': float(np.std([v['accuracy'] for v in results[m][r]])),
                'clean_ratio_mean': float(np.mean([v['clean_ratio'] for v in results[m][r]])),
                'mislabeled_ratio_mean': float(np.mean([v['mislabeled_ratio'] for v in results[m][r]])),
            }
            for r in RETENTION_RATES
        }
        for m in results
    }
}

with open('gais_mm_results_1M.json', 'w') as f:
    json.dump(output, f, indent=2)
print("✅ Saved: gais_mm_results_1M.json")

# ============================================================================
# FINAL SUMMARY
# ============================================================================
elapsed = time.time() - start_time
print("\n" + "="*70)
print("🎉 EXPERIMENT COMPLETE!")
print("="*70)

g_01 = np.mean([x['accuracy'] for x in results['GAIS-MM (Ours)'][0.01]]) * 100
c_01 = np.mean([x['accuracy'] for x in results['CLIP Score'][0.01]]) * 100
g_05 = np.mean([x['accuracy'] for x in results['GAIS-MM (Ours)'][0.05]]) * 100
c_05 = np.mean([x['accuracy'] for x in results['CLIP Score'][0.05]]) * 100

print(f"""
⏱️ Total runtime: {elapsed/60:.1f} minutes ({elapsed/3600:.2f} hours)

📊 DATASET: {n_samples:,} samples, {n_classes} classes, {NOISE_RATE:.0%} noise

🎯 CLIP BLIND SPOT:
   Clean CLIP score:      {clean_clip_mean:.4f}
   Mislabeled CLIP score: {mislabel_clip_mean:.4f}
   → Mislabeled samples have HIGHER CLIP scores!

📈 MAIN RESULTS:
   @ 1% Retention ({int(n_samples * 0.01):,} samples):
      GAIS-MM {g_01:.1f}% vs CLIP {c_01:.1f}% (Δ = {g_01-c_01:+.1f}pp)

   @ 5% Retention ({int(n_samples * 0.05):,} samples):
      GAIS-MM {g_05:.1f}% vs CLIP {c_05:.1f}% (Δ = {g_05-c_05:+.1f}pp)

📁 OUTPUT FILES:
   - gais_mm_results_1M.png/pdf
   - gais_mm_results_1M.json
""")

# Download in Colab
try:
    from google.colab import files
    print("📥 Downloading results...")
    files.download('gais_mm_results_1M.png')
    files.download('gais_mm_results_1M.pdf')
    files.download('gais_mm_results_1M.json')
except:
    print("📁 Files saved in current directory")

print("\n✅ Done! 🎉")
```
